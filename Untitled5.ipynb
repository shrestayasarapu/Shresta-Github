```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator

datagen = ImageDataGenerator(
    rescale=1./255,
    validation_split=0.2
)

train_data = datagen.flow_from_directory(
    '/content/breast_cancer_data',
    target_size=(224, 224),
    batch_size=32,
    class_mode='binary',
    subset='training'
)

val_data = datagen.flow_from_directory(
    '/content/breast_cancer_data',
    target_size=(224, 224),
    batch_size=32,
    class_mode='binary',
    subset='validation'
)
```

```python
from tensorflow.keras import layers, models

model = models.Sequential([
    layers.Conv2D(32, (3,3), activation='relu', input_shape=(224,224,3)),
    layers.MaxPooling2D(2,2),
    layers.Conv2D(64, (3,3), activation='relu'),
    layers.MaxPooling2D(2,2),
    layers.Conv2D(128, (3,3), activation='relu'),
    layers.MaxPooling2D(2,2),
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dropout(0.5),  # helps prevent overfitting
    layers.Dense(1, activation='sigmoid')  # binary output
])

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

history = model.fit(train_data, validation_data=val_data, epochs=10)
```

```python
# Evaluate model
val_loss, val_acc = model.evaluate(val_data)
print(f"Validation Accuracy: {val_acc*100:.2f}%")
print(f"Validation Loss: {val_loss:.4f}")

# Save model
model.save("/content/breast_cancer_model.h5")
print("✅ Model saved as 'breast_cancer_model.h5'")
```

```python
import gradio as gr
import numpy as np
from PIL import Image
import tensorflow as tf

# Load the saved model
model = tf.keras.models.load_model("/content/breast_cancer_model.keras")

# Safe prediction function
def predict(img):
    try:
        # Ensure correct format
        img = img.convert("RGB")
        img = img.resize((224, 224))
        img_array = np.expand_dims(np.array(img) / 255.0, axis=0)

        # Predict
        pred = model.predict(img_array)[0][0]
        confidence = pred if pred > 0.5 else 1 - pred
        label = "🩸 Malignant" if pred > 0.5 else "✅ Benign"

        # Return both label + confidence
        return f"{label}\nConfidence: {confidence*100:.2f}%"

    except Exception as e:
        return f"⚠️ Error processing image: {str(e)}"

# Create Gradio app
demo = gr.Interface(
    fn=predict,
    inputs=gr.Image(type="pil"),
    outputs="text",
    title="Breast Cancer Detection AI",
    description="Upload a breast tissue image (JPG/PNG) to detect if it's benign or malignant.",
    examples=[
        ["/content/Multi Cancer/Multi Cancer/Breast Cancer/breast_benign/Benign (1).jpg"],
        ["/content/Multi Cancer/Multi Cancer/Breast Cancer/breast_malignant/Malignant (1).jpg"]
    ]
)

demo.launch(share=True)
```
